```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Assume df is your DataFrame and is already loaded.
# Separate features and target
X = df.drop(columns=['Pack_Years'])
y = df['Pack_Years']

# Convert categorical variables to numeric (if any)
X = pd.get_dummies(X, drop_first=True)

# Print number of features after one-hot encoding
print(f'Number of features after encoding: {X.shape[1]}')

# Standardize the features
x_scaled = StandardScaler().fit_transform(X)

# Get the number of features to ensure PCA components match
n_features = X.shape[1]  # Update this to the correct number of features

# PCA
pca = PCA(n_components=min(n_features, 30))  # Use the lesser of n_features and 30
pca_features = pca.fit_transform(x_scaled)

# Loadings and explained variance
loadings = pca.components_  # Shape = (n_components, n_features)

# Preparing loadings DataFrame
pc_list = [f'PC{i}' for i in range(1, loadings.shape[0] + 1)]  # Number of components is the first dimension of loadings
loadings_df = pd.DataFrame(loadings.T, columns=pc_list, index=X.columns)

# Show loadings DataFrame
print(loadings_df)

# Plotting Eigenvalues
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(pca.explained_variance_) + 1), pca.explained_variance_, marker='o', linestyle='--')
plt.axhline(y=1, color='r', linestyle='--', label='Kaiser Criterion (1.0)')
plt.ylabel('Eigenvalues')
plt.xlabel('# of Features (Principal Components)')
plt.title('PCA Eigenvalues')
plt.ylim(0, max(pca.explained_variance_) + 1)  # Adding a margin
plt.xticks(range(1, len(pca.explained_variance_) + 1))  # Show ticks for each principal component
plt.grid()
plt.legend()
plt.show()
```
